```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv('/guia_fasecolda.csv')

print("--- Primeras Filas ---")
print(df.head())

print("\n--- Información del Dataset ---")
print(df.info())

print("\n--- Conteo de Valores Faltantes ---")
print(df.isnull().sum())

print("\n--- Estadísticas Descriptivas ---")
print(df.describe())

plt.figure(figsize=(10, 5))
sns.boxplot(x=df['2017'])
plt.title('Distribución de Precios de Vehículos (Año 2017)')
plt.xlabel('Precio (en miles de COP)')
plt.show()

print("\n--- Top 10 Marcas más Comunes ---")
top_marcas = df['Marca'].value_counts().head(10)
print(top_marcas)

plt.figure(figsize=(10, 5))
top_marcas.plot(kind='bar', color='skyblue')
plt.title('Top 10 Marcas más Comunes en el Dataset')
plt.xlabel('Marca')
plt.ylabel('Cantidad de Vehículos')
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Cilindraje', y='2017', alpha=0.6)
plt.title('Relación entre Cilindraje y Precio (2017)')
plt.xlabel('Cilindraje (cc)')
plt.ylabel('Precio (2017)')
plt.show()

precio_por_marca = df.groupby('Marca')['2017'].mean().sort_values(ascending=False).head(10)
print("\n--- Precio Promedio (2017) por Top 10 Marcas ---")
print(precio_por_marca)

peso_por_marca = df.groupby('Marca')['Peso'].mean().sort_values(ascending=False).head(10)
print("\n--- Peso Promedio por Top 10 Marcas ---")
print(peso_por_marca)
```
